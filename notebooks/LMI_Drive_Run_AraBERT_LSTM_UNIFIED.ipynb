## 4) إصلاح AMP/Scheduler في كود التدريب — مع نسخ احتياطية

```python
# AMP scheduler fix: applies to local training code and the persistent Drive source.
#!/usr/bin/env python3
"""Patch incorrect LR scheduler stepping when AMP skips optimizer updates.

Safe/idempotent: backs up changed Python modules, validates syntax before writing.
Both a fast local Colab worktree and a persistent Drive tree may be patched.
This does NOT alter existing weights or claim to retrain models.
"""
from __future__ import annotations

import argparse
import ast
import datetime as dt
import os
import re
from pathlib import Path

MARKER = "AMP_SAFE_SCHEDULER_STEP_V1"
FILES = ("src/training/train_classifier.py", "src/training/train_ner.py")
PATTERN = re.compile(
    r"^(?P<indent>[ \t]*)scaler\.step\(opt\)\r?\n"
    r"(?P=indent)scaler\.update\(\)\r?\n"
    r"(?P=indent)opt\.zero_grad\((?:set_to_none=True)?\)\r?\n"
    r"(?P=indent)sch\.step\(\)", re.MULTILINE,
)


def patch_text(source: str, name: str) -> tuple[str, bool]:
    if MARKER in source:
        return source, False
    matches = list(PATTERN.finditer(source))
    if len(matches) != 1:
        raise RuntimeError(f"{name}: expected exactly one optimizer/scheduler block, found {len(matches)}; not changing file")

    indent = matches[0].group("indent")
    replacement = "\n".join((
        f"{indent}# {MARKER}: schedule only after a real optimizer update",
        f"{indent}scale_before_update = scaler.get_scale()",
        f"{indent}scaler.step(opt)",
        f"{indent}scaler.update()",
        f"{indent}step_skipped_by_amp = bool(amp and scaler.get_scale() < scale_before_update)",
        f"{indent}opt.zero_grad(set_to_none=True)",
        f"{indent}if not step_skipped_by_amp:",
        f"{indent}    optimizer_updates += 1",
        f"{indent}    sch.step()",
        f"{indent}else:",
        f"{indent}    amp_skipped_updates += 1",
    ))
    source = source[:matches[0].start()] + replacement + source[matches[0].end():]

    # Counters reset once per epoch, before the first minibatch is processed.
    epoch_counter = re.compile(r"^(?P<indent>[ \t]*)tot, n = 0\.0, 0[ \t]*$", re.MULTILINE)
    found = list(epoch_counter.finditer(source))
    if len(found) != 1:
        raise RuntimeError(f"{name}: no unique epoch counter initialization; not changing file")
    m = found[0]
    new_counter = m.group(0) + "\n" + m.group("indent") + "optimizer_updates = 0\n" + m.group("indent") + "amp_skipped_updates = 0"
    source = source[:m.start()] + new_counter + source[m.end():]

    # Save AMP diagnostics alongside validation metrics in training_history.csv.
    target = "        hist.append(row)"
    if source.count(target) != 1:
        raise RuntimeError(f"{name}: could not add AMP diagnostics to history")
    source = source.replace(target,
        "        row['optimizer_updates'] = optimizer_updates\n"
        "        row['amp_skipped_updates'] = amp_skipped_updates\n"
        "        hist.append(row)", 1)

    # Do not silently accept an epoch in which no weights could be updated.
    if name == "train_classifier.py":
        entry = "        lg = predict(model, dl_va, dev, amp)"
    else:
        entry = "        model.eval()"
    if source.count(entry) != 1:
        raise RuntimeError(f"{name}: could not insert update-validation guard")
    source = source.replace(entry,
        "        if optimizer_updates == 0:\n"
        "            raise RuntimeError('AMP skipped every optimizer step in this epoch; investigate nonfinite gradients.')\n"
        "        if amp_skipped_updates:\n"
        "            print(f'[AMP] epoch {ep}: skipped {amp_skipped_updates} nonfinite-gradient update(s)', flush=True)\n"
        + entry, 1)
    ast.parse(source, filename=name)
    return source, True


def patch_project(project_root: Path) -> None:
    if not (project_root / "src" / "common.py").exists():
        raise FileNotFoundError(f"Project not found: {project_root}")
    prepared = []
    for rel in FILES:
        p = project_root / rel
        if not p.is_file():
            raise FileNotFoundError(f"Required training source missing: {p}")
        raw = p.read_text(encoding="utf-8")
        edited, changed = patch_text(raw, p.name)
        prepared.append((p, raw, edited, changed))
    # All files validated before any changes are committed.
    for p, original, edited, changed in prepared:
        if not changed:
            print("ALREADY_PATCHED:", p)
            continue
        suffix = dt.datetime.now().strftime("%Y%m%d_%H%M%S_%f")
        bak = p.with_name(p.name + ".before_amp_fix_" + suffix + ".bak")
        bak.write_text(original, encoding="utf-8")
        tmp = p.with_name(p.name + ".tmp_amp_fix")
        tmp.write_text(edited, encoding="utf-8")
        os.replace(tmp, p)
        print("PATCHED:", p)
        print("BACKUP:", bak)
    print("SUCCESS: AMP scheduler is guarded; Python files syntax-checked. Models and datasets untouched.")


def main() -> None:
    ap = argparse.ArgumentParser()
    ap.add_argument("--project-root", default="/content/lmi")
    ap.add_argument("--drive-project", help="Optionally patch persistent project source in Google Drive")
    args = ap.parse_args()
    roots = [Path(args.project_root)]
    if args.drive_project:
        roots.append(Path(args.drive_project))
    for root in dict.fromkeys(roots):
        patch_project(root)



patch_project(Path(ROOT))
patch_project(Path(PROJ))
```

```python
import ast, datetime
SEARCH_MARKER = '.astype(bool)]'

def fix_search_result_masks(project):
    target=Path(project)/'src/integration/engine.py'
    code=target.read_text(encoding='utf-8')
    start=code.find('    def search(self,')
    end=code.find('    def governorates(',start)
    if start<0 or end<0:
        raise RuntimeError('Market.search boundary not found; refusing unsafe change')
    portion=code[start:end]
    old_new=[
        ('u = u[u.specializations.map(lambda s: specialization in split_multi(s))]',
         'u = u.loc[u.specializations.map(lambda s: specialization in split_multi(s)).astype(bool)]'),
        ('u = u[hay.map(lambda h: all(t in h for t in terms))]',
         'u = u.loc[hay.map(lambda h: all(t in h for t in terms)).astype(bool)]')]
    for old,new in old_new:
        if old in portion: portion=portion.replace(old,new,1)
        elif new not in portion: raise RuntimeError('Unexpected search code; not patching '+str(target))
    edited=code[:start]+portion+code[end:]
    ast.parse(edited)
    if edited != code:
        suffix=datetime.datetime.now(datetime.timezone.utc).strftime('%Y%m%d_%H%M%S')
        (target.parent/(target.name+'.before_search_'+suffix+'.bak')).write_text(code,encoding='utf-8')
        target.write_text(edited,encoding='utf-8')
        print('PATCHED:',target)
    else:
        print('ALREADY PATCHED:',target)

for project in (ROOT,PROJ):
    fix_search_result_masks(project)
check_search = """from src.integration.engine import Market
m=Market(); a=m.search('qzxw1998nonexistent',limit=3)
assert len(a)==0 and 'posting_date' in a.columns
b=m.search('',limit=3); assert not b.empty and 'posting_date' in b.columns
print('Search: PASS (empty and populated results)')"""
subprocess.run([sys.executable,'-c',check_search],cwd=ROOT,env=os.environ.copy(),check=True)
```

## 6) تدريب AraBERT لتصنيف نوع الفرصة والنطاق الجغرافي

الدفتر لا يعتمد على وجود ملف `training_config.json` وحده: يتحقق أيضاً من تحميل الأوزان وTokenizer. إذا كان أحدهما تالفاً، يتوقف برسالة بدلاً من تجاوزه. **المهمة التي انقطع تدريبها قبل اكتمالها ستبدأ من الصفر** وفق كود التدريب الحالي.

```python
from transformers import AutoModelForSequenceClassification, AutoModelForTokenClassification, AutoTokenizer
import gc

def verified_training(d, task):
    d=Path(d)
    metadata=d/'training_config.json'
    config=d/'config.json'
    if not metadata.is_file() or not config.is_file():
        return False
    meta=json.loads(metadata.read_text(encoding='utf-8'))
    if meta.get('task')!=task:
        raise ValueError(f'Unexpected training task in {metadata}: {meta.get("task")!r}')
    try:
        tok=AutoTokenizer.from_pretrained(d,local_files_only=True)
        cls = AutoModelForTokenClassification if task=='ner' else AutoModelForSequenceClassification
        mdl=cls.from_pretrained(d,local_files_only=True)
        if mdl.config.num_labels<=1:
            raise ValueError('Model has fewer than two output labels')
        if len(tok) > mdl.get_input_embeddings().weight.shape[0]:
            raise ValueError('Tokenizer vocabulary does not match model embeddings')
        if task != 'ner':
            expected=json.loads((DATA_DIR / f'classification_{task}/label_map.json').read_text(encoding='utf-8'))['id2label']
            actual={str(k):v for k,v in mdl.config.id2label.items()}
            if actual!=expected:
                raise ValueError('Checkpoint id2label does not match current training dataset label_map')
        else:
            tags=json.loads((DATA_DIR/'ner/label_list.json').read_text(encoding='utf-8'))['bio_tags']
            if [mdl.config.id2label[i] for i in range(len(tags))]!=tags:
                raise ValueError('NER checkpoint tags do not match dataset BIO tag list')
        del mdl,tok
        gc.collect()
    except Exception as exc:
        raise RuntimeError(f'Checkpoint metadata exists but model is incomplete/corrupted: {d}. Do not mark it trained. Error: {exc}') from exc
    return True

M=Path(os.environ['LMI_MODELS_DIR'])
for task in ('opportunity_type','geographic_scope'):
    d=M/'classification'/task
    if verified_training(d,task):
        print('VALID TRAINED CHECKPOINT — SKIP:',task)
        continue
    print('TRAINING:',task,flush=True)
    t=time.time()
    subprocess.run([sys.executable,'-m','src.training.train_classifier','--task',task],cwd=ROOT,env=os.environ.copy(),check=True)
    if not verified_training(d,task):
        raise RuntimeError(f'Classifier ended without loadable completion checkpoint: {task}')
    print('COMPLETED:',task,'minutes',round((time.time()-t)/60,1))
```

## 9) فحص بيانات التنبؤ الزمني وإعدادات LSTM

هذه المرحلة **إضافة جديدة** إلى الدفتر: نتحقق من بيانات الوظائف/التدريب وتواريخها، ثم ننشئ أربع عائلات من السلاسل الزمنية الشهرية، ونراجع كفاية البيانات قبل التدريب.

- **المهارات** (skills)، **التخصصات** (specializations)، **المسميات الوظيفية** (job_titles)، **المحافظات** (governorates).
- **المدخلات**: حصة الإعلانات التي تذكر كل عنصر في كل شهر؛ نافذة تاريخية طولها 6 أشهر افتراضياً.
- **التقسيم**: زمني؛ آخر 6 أشهر مكتملة للاختبار، والثلاثة السابقة للتحقق، وما قبلها للتدريب؛ يمنع كود المشروع استخدام أهداف الشهور غير المرصودة أثناء التدريب.
- `LSTM_AS_OF` يثبت تاريخ القطع لضمان إعادة إنتاج نفس الفترة عند تشغيل البيانات الحالية. غيّره فقط إذا حدّثت البيانات وأردت دراسة فترة أحدث.
- `RETRAIN_LSTM = True` يعني أن **Run all ينفذ تدريب LSTM فعلياً** حتى لو وجدت أوزان محفوظة؛ مع نسخ احتياطي قبل نشر المخرجات الجديدة. لتجنب إعادة التدريب في تشغيل لاحق، غيّره إلى `False`.

**تنبيه منهجي:** تقييم Test المتاح في سكربت المشروع هو *تنبؤ خطوة واحدة مع إتاحة القيم الفعلية للأشهر السابقة*، وليس تقييماً مستقلاً لتنبؤ تكراري بستة أشهر. وتوقعات المستقبل ليست حقائق مرصودة.

```python
from datetime import date, datetime
from src.forecasting.series import build_family
from src.forecasting.train_lstm import make_model
from src.common import CONFIG, DATA_DIR

LSTM_FAMILIES = ('skills', 'specializations', 'job_titles', 'governorates')
LSTM_AS_OF = '2026-11-07'   # تاريخ فحص مجموعة البيانات المستخدمة في هذا المشروع
RETRAIN_LSTM = True       # True: إعادة تدريب 4 نماذج فعلية، False: تخطي النموذج السليم
LSTM_CPU_THREADS = 2     # عدد خيوط CPU لسكريبت LSTM (لا يعتمد على GPU في نسخته الحالية)

assert (ROOT / 'src/forecasting/train_lstm.py').is_file(), 'LSTM train script missing'
for f in ('unique_opportunities.csv', 'skills/job_skills.csv'):
    assert (DATA_DIR / f).is_file(), f'Missing forecast source: {DATA_DIR / f}'
cutoff_date = date.fromisoformat(LSTM_AS_OF)
print('LSTM cutoff:', cutoff_date, '| lookback:', CONFIG['lstm']['lookback'],
      '| epochs max:', CONFIG['lstm']['epochs'], '| device: CPU')
for family in LSTM_FAMILIES:
    s = build_family(family, DATA_DIR, as_of=cutoff_date)
    if s.get('status') != 'ok':
        raise RuntimeError(f'{family}: insufficient data for training: {s.get("status")}')
    n_months = len(s['series'])
    n_train = n_months - CONFIG['lstm']['val_months'] - CONFIG['lstm']['test_months']
    print(f'{family:17s} | items={s["series"].shape[1]:3d} | months={n_months:3d} | '
          f'train={n_train} | val={CONFIG["lstm"]["val_months"]} | test={CONFIG["lstm"]["test_months"]}'
          f' | last month={s["series"].index[-1]}')
print('LSTM data preflight: PASS')
```

## 10) تدريب نماذج LSTM الأربعة وحفظها في Drive

**هنا يحدث تدريب LSTM الفعلي.** تشغّل الخلية `python -m src.forecasting.train_lstm --families ...` لكل عائلة باستخدام كود PyTorch الأصلي، مع تنفيذ **تدريب + اختيار المعمارية على Validation + اختبار Test + مقارنة Naive وMoving Average**.

لحماية الملفات السابقة: يدرب أولاً في مساحة مؤقتة داخل Colab، ويتأكد من وجود الأوزان وسلامتها ومن صحة مقاييس التقييم، **ثم** يحتفظ بنسخة احتياطية في `MyDrive/LMI_PROJECT/LMI_LSTM_BACKUPS/` وينشر الأوزان والتقارير داخل `models_dl/lstm/` و`reports/evaluation/` في Drive. إذا لم تكن بيانات إحدى العائلات كافية، تتوقف الخلية ولا تصف المهمة بأنها مكتملة.

⚠️ لا توقف الخلية في منتصف نسخ الملفات إلى Drive. بعد اكتمال كل عائلة تُحفظ نتيجتها؛ وعند إعادة تشغيل الدفتر تستطيع تغيير `RETRAIN_LSTM` إلى `False` للاستفادة من العائلات التي اكتملت.

```python
import os, sys, json, time, shutil, subprocess
import numpy as np
import pandas as pd
import torch
from pathlib import Path

LSTM_MODELS = Path(os.environ['LMI_MODELS_DIR']) / 'lstm'
LSTM_EVAL = Path(os.environ['LMI_REPORTS_DIR']) / 'evaluation'
LSTM_STAGE = ROOT / '.lstm_training_stage'
LSTM_BACKUPS = PROJ / 'LMI_LSTM_BACKUPS'
LSTM_RUN_ID = datetime.now().strftime('%Y%m%d_%H%M%S')

def verify_lstm_artifacts(family, model_folder, evaluation_folder):
    """Validates real PyTorch weights and recomputes held-out scores from saved predictions."""
    model_folder, evaluation_folder = Path(model_folder), Path(evaluation_folder)
    needed = [model_folder/'lstm.pt', model_folder/'config.json', model_folder/'scaler.json',
              evaluation_folder/'metrics.json', evaluation_folder/'test_predictions.csv',
              evaluation_folder/'training_history.csv', evaluation_folder/'future_forecast.csv']
    missing = [str(f) for f in needed if not f.is_file() or f.stat().st_size <= 0]
    if missing:
        raise RuntimeError(f'{family}: incomplete model/report files: {missing}')
    conf = json.loads((model_folder/'config.json').read_text(encoding='utf-8'))
    metrics = json.loads((evaluation_folder/'metrics.json').read_text(encoding='utf-8'))
    assert conf.get('family') == family and metrics.get('trained') is True, f'{family}: metadata mismatch'
    assert conf['test_period'][0] > conf['validation_period'][-1] > conf['train_period'][-1], f'{family}: time leakage in split'
    net = make_model(int(conf['hidden_size']), int(CONFIG['lstm']['num_layers']), float(conf['dropout']))
    state = torch.load(model_folder/'lstm.pt', map_location='cpu', weights_only=True)
    net.load_state_dict(state, strict=True)
    net.eval()
    with torch.no_grad():
        sample = net(torch.zeros((2, int(conf['lookback']), 1), dtype=torch.float32)).numpy()
    if sample.shape != (2,) or not np.isfinite(sample).all():
        raise RuntimeError(f'{family}: model inference returned invalid values')
    pr = pd.read_csv(evaluation_folder/'test_predictions.csv')
    required = {'month', 'item', 'actual', 'lstm', 'naive', 'moving_avg_3'}
    if not required.issubset(pr.columns) or pr.empty:
        raise RuntimeError(f'{family}: missing forecast predictions/columns')
    assert pr.month.min() >= conf['test_period'][0] and pr.month.max() <= conf['test_period'][-1], f'{family}: test dates incorrect'
    observed = pr.dropna(subset=['actual'])
    if observed.empty:
        raise RuntimeError(f'{family}: no observed ground truth to evaluate')
    for name in ('lstm', 'naive', 'moving_avg_3'):
        a = observed['actual'].to_numpy(dtype=float)
        p = observed[name].to_numpy(dtype=float)
        good = np.isfinite(a) & np.isfinite(p)
        if not good.any():
            raise RuntimeError(f'{family}: non-finite scores for {name}')
        mae = float(np.mean(np.abs(a[good] - p[good])))
        rmse = float(np.sqrt(np.mean((a[good] - p[good])**2)))
        if abs(mae - float(metrics['test'][name]['mae'])) > 2e-5 or abs(rmse - float(metrics['test'][name]['rmse'])) > 2e-5:
            raise RuntimeError(f'{family}: metrics.json does not match test_predictions.csv for {name}')
    del net, state
    return metrics

def publish_lstm_family(family, model_src, eval_src):
    """Back up both folders and restore them if publishing an artifact fails."""
    targets = [(Path(model_src), LSTM_MODELS / family, 'model'),
               (Path(eval_src), LSTM_EVAL / ('lstm_' + family), 'evaluation')]
    previous = []
    for _, dst, kind in targets:
        backup = LSTM_BACKUPS / LSTM_RUN_ID / family / kind
        existed = dst.exists()
        if existed:
            backup.parent.mkdir(parents=True, exist_ok=True)
            shutil.copytree(dst, backup, dirs_exist_ok=True)
        previous.append((dst, backup, existed))
    try:
        for src, dst, _ in targets:
            dst.parent.mkdir(parents=True, exist_ok=True)
            if dst.exists():
                shutil.rmtree(dst)
            shutil.copytree(src, dst)
        verify_lstm_artifacts(family, targets[0][1], targets[1][1])
    except Exception:
        for dst, backup, existed in previous:
            if dst.exists():
                shutil.rmtree(dst)
            if existed and backup.exists():
                shutil.copytree(backup, dst)
        raise

start_lstm = time.time()
for family in LSTM_FAMILIES:
    current_model = LSTM_MODELS / family
    current_report = LSTM_EVAL / ('lstm_' + family)
    if not RETRAIN_LSTM:
        try:
            old_results = verify_lstm_artifacts(family, current_model, current_report)
        except Exception as e:
            print(f'RETRAIN REQUIRED for {family}: {e}', flush=True)
        else:
            print(f'SKIP verified LSTM: {family}; MAE={old_results["test"]["lstm"]["mae"]}', flush=True)
            continue
    print('TRAINING LSTM:', family, '| using CPU', flush=True)
    t0 = time.time()
    stage_models = LSTM_STAGE / family / 'models_dl'
    stage_reports = LSTM_STAGE / family / 'reports'
    if stage_models.parent.exists():
        shutil.rmtree(stage_models.parent)
    stage_models.mkdir(parents=True)
    stage_reports.mkdir(parents=True)
    env = os.environ.copy()
    env['LMI_MODELS_DIR'] = str(stage_models)
    env['LMI_REPORTS_DIR'] = str(stage_reports)
    cmd = [sys.executable, '-m', 'src.forecasting.train_lstm',
           '--families', family, '--as-of', LSTM_AS_OF, '--cpu-threads', str(LSTM_CPU_THREADS)]
    subprocess.run(cmd, cwd=ROOT, env=env, check=True)
    s_model = stage_models / 'lstm' / family
    s_report = stage_reports / 'evaluation' / ('lstm_' + family)
    result = verify_lstm_artifacts(family, s_model, s_report)
    publish_lstm_family(family, s_model, s_report)
    print(f'SAVED to Drive: {family} | Test MAE={result["test"]["lstm"]["mae"]:.6f} | '
          f'Naive={result["test"]["naive"]["mae"]:.6f} | MA-3={result["test"]["moving_avg_3"]["mae"]:.6f} '
          f'| elapsed={(time.time()-t0)/60:.1f} min', flush=True)

# Merge new families with any existing summary without destroying other reports.
summary_file = LSTM_EVAL / 'lstm_summary.json'
old_summary = json.loads(summary_file.read_text(encoding='utf-8')) if summary_file.is_file() else {}
for family in LSTM_FAMILIES:
    r = verify_lstm_artifacts(family, LSTM_MODELS/family, LSTM_EVAL/('lstm_' + family))
    old_summary[family] = {k: r.get(k) for k in
        ('trained', 'reason', 'series', 'test', 'test_period', 'lstm_beats_naive_mae', 'lstm_beats_ma3_mae')}
summary_file.parent.mkdir(parents=True, exist_ok=True)
tmp_summary = summary_file.with_suffix('.tmp')
tmp_summary.write_text(json.dumps(old_summary, ensure_ascii=False, indent=2), encoding='utf-8')
os.replace(tmp_summary, summary_file)
print(f'LSTM stage complete: 4/4 verified; elapsed {(time.time()-start_lstm)/60:.1f} minutes')
print('Saved in Drive:', LSTM_MODELS, 'and', LSTM_EVAL)
```

## 11) تقييم LSTM النهائي: MAE وRMSE والرسوم البيانية

تقرأ هذه الخلية **تقارير Test الفعلية التي أنشأها سكربت التدريب**، وتعيد التحقق من قيم MAE وRMSE مباشرة من التنبؤات المحفوظة، ثم تعرض جدول مقارنة لكل عائلة ورسوماً بيانية للنتائج.

- **MAE وRMSE الأصغر أفضل**. إذا تفوق Naive أو Moving Average على LSTM فلا نخفي ذلك.
- نتائج الاختبار هنا تخص التنبؤ لشهر واحد مع إدخال الأشهر السابقة الحقيقية، ولا تقيس بشكل مستقل أفق التنبؤ التكراري لستة أشهر.
- ملفات `future_forecast.csv` عبارة عن **توقعات مستقبلية** لا عن بيانات مرصودة، فلا تحسب لها دقة Test.
- يُحفظ جدول ملخص قابل للعرض في `reports/evaluation/lstm_comparison_notebook.csv` والرسم في PNG.

```python
import matplotlib.pyplot as plt
from IPython.display import display

rows=[]
for family in LSTM_FAMILIES:
    metrics=verify_lstm_artifacts(family, LSTM_MODELS/family, LSTM_EVAL/('lstm_'+family))
    t=metrics['test']
    choices={k: t[k]['mae'] for k in ('lstm', 'naive', 'moving_avg_3')}
    rows.append({'model': family, 'series': metrics['series'],
                 'test_period': ' → '.join(metrics['test_period']),
                 'test_samples': t['lstm']['n'],
                 'LSTM_MAE': t['lstm']['mae'], 'LSTM_RMSE': t['lstm']['rmse'],
                 'Naive_MAE': t['naive']['mae'], 'MovingAvg3_MAE': t['moving_avg_3']['mae'],
                 'best_method': min(choices, key=choices.get)})
result_table=pd.DataFrame(rows)
display(result_table)
out_csv=LSTM_EVAL/'lstm_comparison_notebook.csv'
result_table.to_csv(out_csv, index=False, encoding='utf-8-sig')

fig,ax=plt.subplots(figsize=(10,4.5))
x=np.arange(len(result_table)); width=0.25
for i,(key,label) in enumerate([('LSTM_MAE','LSTM'),('Naive_MAE','Naive'),('MovingAvg3_MAE','Moving avg. 3')]):
    ax.bar(x+(i-1)*width, result_table[key].astype(float), width=width, label=label)
ax.set_xticks(x, result_table['model'], rotation=13)
ax.set_ylabel('MAE on held-out test (lower is better)')
ax.set_title('LMI — LSTM vs simple forecasting baselines')
ax.legend(); ax.grid(axis='y',alpha=.25)
fig.tight_layout()
out_png=LSTM_EVAL/'lstm_comparison_notebook.png'
fig.savefig(out_png, dpi=160)
plt.show()
print('Saved evaluation:',out_csv,'|',out_png)

# Demonstrate that the saved model and real test metric artifacts can be reopened.
for family in LSTM_FAMILIES:
    cf=json.loads((LSTM_MODELS/family/'config.json').read_text(encoding='utf-8'))
    print(f'{family}: LSTM({cf["hidden_size"]} hidden units) | '
          f'train={cf["train_period"]} validation={cf["validation_period"]} test={cf["test_period"]}')
print('LSTM evaluation: PASS (4 families, weights loadable, metrics recalculated)')
```

## 14) تشغيل Streamlit وفحص استجابة الخادم قبل مشاركة الرابط

الرابط المؤقت عام: لا تدخل بيانات شخصية أو سرية فيه. تحفظ سجلات الخادم للمراجعة في `/content/lmi_streamlit.log`. أوقف الخلية لإيقاف التطبيق.

```python
import urllib.request, re, threading, queue
cloud_bin=Path('/content/cloudflared')
if not cloud_bin.exists():
    url='https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64'
    urllib.request.urlretrieve(url,cloud_bin)
    cloud_bin.chmod(0o755)
logpath=Path('/content/lmi_streamlit.log')
app_log=logpath.open('w',encoding='utf-8')
app=subprocess.Popen([sys.executable,'-m','streamlit','run','app_dl/main.py','--server.port','8501','--server.headless','true'],
                     cwd=ROOT,env=os.environ.copy(),stdout=app_log,stderr=subprocess.STDOUT)
try:
    status=None
    for _ in range(40):
        if app.poll() is not None:
            raise RuntimeError('Streamlit exited; check /content/lmi_streamlit.log:\n'+logpath.read_text(encoding='utf-8')[-3500:])
        try:
            with urllib.request.urlopen('http://127.0.0.1:8501/_stcore/health',timeout=1) as response:
                if response.status==200:
                    status=200;break
        except Exception:
            time.sleep(1)
    if status!=200:
        raise RuntimeError('Streamlit did not respond; inspect /content/lmi_streamlit.log')
    print('Streamlit HTTP health: PASS')
    tun=subprocess.Popen([str(cloud_bin),'tunnel','--url','http://localhost:8501'],
                         stdout=subprocess.PIPE,stderr=subprocess.STDOUT,text=True,bufsize=1)
    lines=queue.Queue()
    def reader():
        for line in tun.stdout:
            lines.put(line)
    threading.Thread(target=reader,daemon=True).start()
    public=None; deadline=time.monotonic()+75
    while time.monotonic()<deadline:
        if tun.poll() is not None and lines.empty():
            raise RuntimeError('cloudflared terminated before providing a link')
        try:
            line=lines.get(timeout=1)
        except queue.Empty:
            continue
        match=re.search(r'https://[\w-]+\.trycloudflare\.com',line)
        if match:
            public=match.group(0);break
    if not public:
        raise RuntimeError('No Cloudflare URL within 75s; inspect internet connectivity and tunnel service')
    print('App link:',public)
    while app.poll() is None and tun.poll() is None:
        time.sleep(5)
finally:
    for proc in (locals().get('tun'),app):
        if proc is not None and proc.poll() is None:
            proc.terminate()
    app_log.close()
    print('Streamlit log:',logpath)
```
